```python
def fetch_dataset(dx=80, dy=80, dimx=45, dimy=45):

    df_attrs = pd.read_csv(ATTRIBUTES_PATH, sep='\t', skiprows=1,)
    df_attrs = pd.DataFrame(df_attrs.iloc[:, :-1].values, columns = df_attrs.columns[1:])

    photo_ids = []
    for dirpath, dirnames, filenames in os.walk(DATASET_PATH):
        for fname in filenames:
            if fname.endswith(".jpg"):
                fpath = os.path.join(dirpath, fname)
                photo_id = fname[:-4].replace('_', ' ').split()
                person_id = ' '.join(photo_id[:-1])
                photo_number = int(photo_id[-1])
                photo_ids.append({'person': person_id, 'imagenum': photo_number, 'photo_path': fpath})

    photo_ids = pd.DataFrame(photo_ids)
    df = pd.merge(df_attrs, photo_ids, on=('person', 'imagenum'))

    assert len(df) == len(df_attrs), "Потеряны данные при объединении датафреймов!"

    images = df['photo_path'].apply(imageio.imread)\
                                .apply(lambda img:img[dy:-dy, dx:-dx])\
                                .apply(lambda img: np.array(Image.fromarray(img).resize([dimx, dimy])) )

    images = np.stack(images.values).astype('uint8')
    attrs = df.drop(["photo_path","person","imagenum"],axis=1)

    return images, attrs
```

Реализуем autoencoder. Архитектуру (conv, fully-connected, ReLu, etc) можете выбирать сами. Экспериментируйте!

```python
class Autoencoder(torch.nn.Module):
  def __init__(self):
    super().__init__()
    # изображения имеют размер 45 * 45
    self.encoder = nn.Sequential(
        nn.Conv2d(3, 40, kernel_size=4, stride=1, padding=1),
        nn.ReLU(),
        nn.Conv2d(40, 32, kernel_size=4, stride=2, padding=1),
        nn.ReLU(),
        nn.Conv2d(32, 16, kernel_size=4, stride=2, padding=1),
        nn.ReLU(),
    )
    self.decoder = nn.Sequential(
        nn.ConvTranspose2d(16, 16, kernel_size=4, stride=2, padding=1, output_padding=1), # output_padding служит просто для того, чтобы входной и выходной размер картинки совпали
        nn.ReLU(),
        nn.ConvTranspose2d(16, 32, kernel_size=4, stride=2, padding=1),
        nn.ReLU(),
        nn.ConvTranspose2d(32, 40, kernel_size=4, stride=1, padding=1),
        nn.ReLU(),
        nn.Conv2d(40, 3, kernel_size=3, stride=1, padding=0),
    )

  def forward(self, sample):
    latent = self.encoder(sample)
    reconstructed = self.decoder(latent)
    return reconstructed, latent
```

```python
criterion = F.mse_loss

device = 'cuda:0' if torch.cuda.is_available() else 'cpu'

print(device)
autoencoder = Autoencoder().to(device)

optimizer = torch.optim.Adam(autoencoder.parameters(), lr=0.001)
```

## 1.3 Обучение (2 балла)

Осталось написать код обучения автоэнкодера. При этом было бы неплохо в процессе иногда смотреть, как автоэнкодер реконструирует изображения на данном этапе обучения. Наример, после каждой эпохи (прогона train выборки через автоэнкодер) можно смотреть, какие реконструкции получились для каких-то изображений val выборки.

А, ну еще было бы неплохо выводить графики train и val лоссов в процессе тренировки =)

```python
train_losses = []
val_losses = []
n_epochs = 20
lst_reconst = []
last_reconst = []

for epoch in tqdm_notebook(range(n_epochs)):
    autoencoder.train()
    train_losses_per_epoch = []
    for i, X_batch in enumerate(train_loader):
        optimizer.zero_grad()
        reconstructed, latent = autoencoder(X_batch.to(device))
        loss = criterion(reconstructed, X_batch.to(device))
        loss.backward()
        optimizer.step()
        train_losses_per_epoch.append(loss.item())

    train_losses.append(np.mean(train_losses_per_epoch))

    autoencoder.eval()
    val_losses_per_epoch = []
    it = 0
    cur_rec = [] # храним 5 реконструированных батчек на текущей эпохе, то есть 5 * 4 = 20 картинок
    with torch.no_grad():
        for X_batch in val_loader:
          reconstructed, latent = autoencoder(X_batch.to(device))
          if it <= 4:
            cur_rec.append(reconstructed)
            if epoch == n_epochs - 1: # проверяем как автоэнкодер декодирует картинки после последней эпохи(будучи натренированным)
              last_reconst.append(reconstructed)
          loss = criterion(reconstructed, X_batch.to(device))
          val_losses_per_epoch.append(loss.item())
          it += 1
    lst_reconst.append(cur_rec)
    val_losses.append(np.mean(val_losses_per_epoch))
```

```python
plt.figure(figsize=(10, 7))
plt.plot(np.arange(len(train_losses)), train_losses, label='Train')
plt.plot(np.arange(len(val_losses)), val_losses, label='Validation')

plt.ylim(0, 0.05)
plt.xlabel('Epoch')
plt.title('MSE loss')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import torch
import numpy as np

# Функция для отрисовки эпохи (20 картинок)
def show_epoch_reconstructions(rec_batch_list, epoch_num):
    """
    rec_batch_list: список из 5 батчей (каждый батч [4, 3, 45, 45])
    epoch_num: номер эпохи для заголовка
    """
    plt.figure(figsize=(15, 8))

    for batch_idx, batch_tensor in enumerate(rec_batch_list):
        # batch_tensor имеет размер [4, 3, 45, 45]
        # Проходимся по 4 картинкам внутри батча
        for img_idx, tensor in enumerate(batch_tensor):
            plt.subplot(5, 4, batch_idx * 4 + img_idx + 1)

            # Преобразуем (C, H, W) -> (H, W, C) для matplotlib
            img = tensor.cpu().permute(1, 2, 0).numpy()

            img = np.clip(img, 0, 1)

            plt.imshow(img)
            plt.axis('off')

    plt.suptitle(f"Epoch {epoch_num + 1}", fontsize=16)
    plt.tight_layout()
    plt.show()

# Функция для финального результата (тоже 20 картинок)
def show_final_reconstructions(rec_batch_list):
    """
    Отображает финальные реконструкции
    """
    plt.figure(figsize=(15, 8))

    for batch_idx, batch_tensor in enumerate(rec_batch_list):
        for img_idx, tensor in enumerate(batch_tensor):
            plt.subplot(5, 4, batch_idx * 4 + img_idx + 1)
            img = tensor.cpu().permute(1, 2, 0).numpy()
            img = np.clip(img, 0, 1)
            plt.imshow(img)
            plt.axis('off')

    plt.suptitle("Последняя эпоха", fontsize=16)
    plt.tight_layout()
    plt.show()


# 1. Посмотреть динамику на разных этапах
print("Показываю эволюцию обучения...")
# Покажем результат на нескольких эпохах
epochs_to_show = [0, 5, 10, 15]

for epoch in epochs_to_show:
    show_epoch_reconstructions(lst_reconst[epoch], epoch)

# 2. Отдельно финальные картинки
print("Показываю финальный результат...")
# last_reconst содержит те же 5 батчей, что и lst_reconst[24]
show_final_reconstructions(last_reconst)
```

### Архитектура

На картинке ниже представлена архитектура простого Conditional VAE.
